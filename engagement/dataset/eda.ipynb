```python
import sqlite3
import pandas as pd

# Connect to the SQLite database
conn = sqlite3.connect('tweets.db')

# Query the tweet_observations table and load it into a pandas DataFrame
query = "SELECT * FROM tweet_observations"
tweet_df = pd.read_sql_query(query, conn)

# Display the first few rows of the dataframe
tweet_df.head()
```

```python
# group by tweet_id and count the number of observations
grouped = tweet_df.groupby('tweet_id').size().reset_index(name='count')
# sort by count in descending order
grouped = grouped.sort_values(by='count', ascending=False)
grouped.head()
```
